```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.preprocessing import MinMaxScaler, StandardScaler
```

```python
df = pd.read_csv('mining_distribution_corrected_v2.csv')

df.info()
```

```python
df = df.drop(columns=["Date"])

weather_order = {'Light Rain': 0, 'Cloudy': 1, 'Sunny': 2}
df['Weather_Encoded'] = df['Weather'].map(weather_order)

df.drop(columns=['Weather'], inplace=True)

df.head(10)
```

```python
X = df[['Stock', 'Transport Capacity', 'Loading Time', 'Weather_Encoded']]
Y = df['Shipping Volume']
```

```python
for col in X.columns:
    plt.scatter(X[col], Y)
    plt.title(col)
    plt.show()
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, Y, test_size=0.2, random_state=621)

X_scaler = StandardScaler()
X_scaler.fit(X_train)

Y_scaler = StandardScaler()
Y_scaler.fit(y_train.values.reshape(-1, 1))

X_train_scaled = X_scaler.transform(X_train)
X_test_scaled = X_scaler.transform(X_test)

y_train_scaled = Y_scaler.transform(y_train.values.reshape(-1, 1))
y_test_scaled = Y_scaler.transform(y_test.values.reshape(-1, 1))
```

```python
model = LinearRegression()
model.fit(X_train_scaled, y_train_scaled)

y_pred_scaled = model.predict(X_test_scaled)

y_pred = Y_scaler.inverse_transform(y_pred_scaled)

mae = mean_absolute_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print("Mean Squared Error:", mae)
print("R-squared:", r2)
```

# SVM Scaled

```python
from sklearn import svm

model = svm.SVR(kernel='poly')
model.fit(X_train_scaled, y_train_scaled.ravel())

y_pred_scaled = model.predict(X_test_scaled)
y_pred = Y_scaler.inverse_transform(y_pred_scaled.reshape(-1, 1))

mae = mean_absolute_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print("Mean Absolute Error:", mae)
print("R-squared:", r2)

test_predict = model.predict(X_scaler.transform([[24, 29, 35, 22]]))

print(Y_scaler.inverse_transform(test_predict.reshape(-1, 1)))
```

# SVM Unscaled

```python
model = svm.SVR(kernel='linear')
model.fit(X_train, y_train)

y_pred = model.predict(X_test)

mae = mean_absolute_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print("Mean Absolute Error:", mae)
print("R-squared:", r2)

test_predict = model.predict([[120, 10, 8, 2]])

print(test_predict)
```

```python
from sklearn.ensemble import RandomForestRegressor

model = RandomForestRegressor()
model.fit(X_train, y_train)

y_pred = model.predict(X_test)

mae = mean_absolute_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print("Mean Squared Error:", mae)
print("R-squared:", r2)
```

```python
from sklearn.ensemble import GradientBoostingRegressor

model = GradientBoostingRegressor()
model.fit(X_train, y_train)

y_pred = model.predict(X_test)

mae = mean_absolute_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print("Mean Squared Error:", mae)
print("R-squared:", r2)
```

# Polynomial Regression

```python
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.pipeline import Pipeline

model = Pipeline([('poly', PolynomialFeatures(degree=2)),
                  ('linear', LinearRegression(fit_intercept=False))])

model.fit(X_train, y_train)

y_pred = model.predict(X_test)

mae = mean_absolute_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print("Mean Absolute Error:", mae)
print("R Squared:", r2)

test_predict = model.predict([[120, 10, 26, 1]])

print(test_predict)
```

```python
import pickle

with open('shipping.pkl', 'wb') as f:
    pickle.dump(model, f)
```
